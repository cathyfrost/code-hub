```python
import matplotlib.pyplot as plt
import numpy as np

# ══════════════════════════════════════════
#  数据（来自 04_kmeans_clustering.ipynb）
# ══════════════════════════════════════════

k_range = list(range(2, 21))

wcss = [5905.95, 5888.67, 5876.78, 5865.11, 5868.35, 5847.85, 5817.85,
        5821.95, 5811.79, 5792.43, 5800.22, 5787.94, 5773.18, 5759.93,
        5756.89, 5736.45, 5720.61, 5722.78, 5723.85]

sil = [0.0148, 0.0149, 0.0147, 0.0137, 0.0144, 0.0147, 0.0164,
       0.0157, 0.0165, 0.0174, 0.0159, 0.0167, 0.0174, 0.0181,
       0.0182, 0.0200, 0.0204, 0.0202, 0.0194]

ch = [44.4, 31.3, 25.1, 21.9, 16.8, 17.7, 19.8,
      16.7, 16.1, 16.6, 14.3, 14.2, 14.4, 14.4,
      13.7, 14.2, 14.4, 13.5, 12.7]

db = [4.1983, 5.7561, 6.7579, 9.8806, 6.5836, 8.2766, 7.6897,
      8.0254, 7.9921, 7.8336, 7.9601, 7.4152, 7.0798, 7.0010,
      7.2493, 6.6481, 6.4893, 6.5124, 6.8655]

# ══════════════════════════════════════════
#  绘图
# ══════════════════════════════════════════

plt.rcParams["font.sans-serif"] = ["SimHei", "Microsoft YaHei", "Arial Unicode MS"]
plt.rcParams["axes.unicode_minus"] = False

fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle("K 值搜索过程中各评估指标变化趋势", fontsize=16, fontweight="bold", y=0.97)

colors = ["#1565c0", "#43a047", "#ef6c00", "#c62828"]
best_k = 10
best_idx = k_range.index(best_k)

# (a) WCSS 肘部法则
ax = axes[0, 0]
ax.plot(k_range, wcss, "o-", color=colors[0], linewidth=2, markersize=5, zorder=3)
ax.scatter([best_k], [wcss[best_idx]], color=colors[3], s=120, zorder=5, edgecolors="white", linewidths=2)
ax.annotate(f"K={best_k}\nWCSS={wcss[best_idx]:.1f}", xy=(best_k, wcss[best_idx]),
            xytext=(best_k + 2, wcss[best_idx] + 30), fontsize=9,
            arrowprops=dict(arrowstyle="->", color=colors[3], lw=1.5), color=colors[3], fontweight="bold")
ax.set_title("(a) WCSS 肘部法则", fontsize=12, fontweight="bold", pad=10)
ax.set_xlabel("K 值", fontsize=10)
ax.set_ylabel("WCSS", fontsize=10)
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3, linestyle="--")
ax.spines[["top", "right"]].set_visible(False)

# (b) 轮廓系数
ax = axes[0, 1]
ax.plot(k_range, sil, "o-", color=colors[1], linewidth=2, markersize=5, zorder=3)
ax.scatter([best_k], [sil[best_idx]], color=colors[3], s=120, zorder=5, edgecolors="white", linewidths=2)
ax.annotate(f"K={best_k}\nSil={sil[best_idx]:.4f}", xy=(best_k, sil[best_idx]),
            xytext=(best_k + 2, sil[best_idx] + 0.002), fontsize=9,
            arrowprops=dict(arrowstyle="->", color=colors[3], lw=1.5), color=colors[3], fontweight="bold")
ax.set_title("(b) 轮廓系数 (Silhouette Score)", fontsize=12, fontweight="bold", pad=10)
ax.set_xlabel("K 值", fontsize=10)
ax.set_ylabel("轮廓系数", fontsize=10)
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3, linestyle="--")
ax.spines[["top", "right"]].set_visible(False)

# (c) CH 指数
ax = axes[1, 0]
ax.plot(k_range, ch, "o-", color=colors[2], linewidth=2, markersize=5, zorder=3)
ax.scatter([best_k], [ch[best_idx]], color=colors[3], s=120, zorder=5, edgecolors="white", linewidths=2)
ax.annotate(f"K={best_k}\nCH={ch[best_idx]:.1f}", xy=(best_k, ch[best_idx]),
            xytext=(best_k + 2, ch[best_idx] + 5), fontsize=9,
            arrowprops=dict(arrowstyle="->", color=colors[3], lw=1.5), color=colors[3], fontweight="bold")
ax.set_title("(c) Calinski-Harabasz 指数", fontsize=12, fontweight="bold", pad=10)
ax.set_xlabel("K 值", fontsize=10)
ax.set_ylabel("CH 指数", fontsize=10)
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3, linestyle="--")
ax.spines[["top", "right"]].set_visible(False)

# (d) DB 指数
ax = axes[1, 1]
ax.plot(k_range, db, "o-", color="#7b1fa2", linewidth=2, markersize=5, zorder=3)
ax.scatter([best_k], [db[best_idx]], color=colors[3], s=120, zorder=5, edgecolors="white", linewidths=2)
ax.annotate(f"K={best_k}\nDB={db[best_idx]:.4f}", xy=(best_k, db[best_idx]),
            xytext=(best_k + 2, db[best_idx] + 0.5), fontsize=9,
            arrowprops=dict(arrowstyle="->", color=colors[3], lw=1.5), color=colors[3], fontweight="bold")
ax.set_title("(d) Davies-Bouldin 指数", fontsize=12, fontweight="bold", pad=10)
ax.set_xlabel("K 值", fontsize=10)
ax.set_ylabel("DB 指数", fontsize=10)
ax.set_xticks(k_range)
ax.grid(True, alpha=0.3, linestyle="--")
ax.spines[["top", "right"]].set_visible(False)

plt.tight_layout(rect=[0, 0, 1, 0.94])
plt.savefig("k_value_search_metrics.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
print("已保存: k_value_search_metrics.png")
```

```python
import matplotlib.pyplot as plt
import numpy as np

# ══════════════════════════════════════════
#  真实数据（来自 05_junk_classifier.ipynb）
# ══════════════════════════════════════════

# MultinomialNB 混淆矩阵
# 技术帖(0) 511条, 非技术帖(1) 1035条
# precision(技术)=0.9577, recall(技术)=0.8865
# precision(非技术)=0.9459, recall(非技术)=0.9807
# TP(非技术正确)=1035*0.9807≈1015, FN(非技术漏判)=1035-1015=20
# TN(技术正确)=511*0.8865≈453, FP(技术误判)=511-453=58
nb_cm = np.array([
    [453, 58],    # 真实=技术帖: 正确453, 误判为非技术58
    [20, 1015],   # 真实=非技术帖: 误判为技术20, 正确1015
])

# LinearSVC 混淆矩阵
# precision(技术)=0.94, recall(技术)=0.91
# precision(非技术)=0.952, recall(非技术)=0.97
# TN(技术正确)=511*0.91≈465, FP(技术误判)=511-465=46
# TP(非技术正确)=1035*0.97≈1004, FN(非技术漏判)=1035-1004=31
svc_cm = np.array([
    [465, 46],    # 真实=技术帖: 正确465, 误判为非技术46
    [31, 1004],   # 真实=非技术帖: 误判为技术31, 正确1004
])

# ══════════════════════════════════════════
#  绘图
# ══════════════════════════════════════════

plt.rcParams["font.sans-serif"] = ["SimHei", "Microsoft YaHei", "Arial Unicode MS"]
plt.rcParams["axes.unicode_minus"] = False

fig, axes = plt.subplots(1, 2, figsize=(14, 5.5))
fig.suptitle("非技术内容分类器混淆矩阵对比", fontsize=15, fontweight="bold", y=1.02)

labels = ["技术帖 (0)", "非技术帖 (1)"]
cms = [nb_cm, svc_cm]
titles = [
    "MultinomialNB\n(Accuracy=0.9495, F1=0.9630)",
    "LinearSVC\n(Accuracy=0.9483, F1=0.9616)",
]

for idx, (ax, cm, title) in enumerate(zip(axes, cms, titles)):
    # 绘制热力图
    im = ax.imshow(cm, interpolation="nearest", cmap="Blues", aspect="auto")

    # 在格子中标注数值
    thresh = cm.max() / 2.0
    for i in range(2):
        for j in range(2):
            val = cm[i, j]
            color = "white" if val > thresh else "black"
            # 同时显示数量和百分比
            total_row = cm[i].sum()
            pct = val / total_row * 100
            ax.text(j, i, f"{val}\n({pct:.1f}%)",
                    ha="center", va="center", color=color,
                    fontsize=13, fontweight="bold")

    ax.set_xticks([0, 1])
    ax.set_yticks([0, 1])
    ax.set_xticklabels(labels, fontsize=11)
    ax.set_yticklabels(labels, fontsize=11)
    ax.set_xlabel("预测类别", fontsize=12, labelpad=8)
    ax.set_ylabel("真实类别", fontsize=12, labelpad=8)
    ax.set_title(title, fontsize=12, fontweight="bold", pad=12)

    # 添加颜色条
    cbar = plt.colorbar(im, ax=ax, fraction=0.046, pad=0.04)
    cbar.ax.tick_params(labelsize=9)

plt.tight_layout()
plt.savefig("confusion_matrix_comparison.png", dpi=300, bbox_inches="tight", facecolor="white")
plt.show()
print("已保存: confusion_matrix_comparison.png")
```
